```python
import spacy
```

```python
nlp = spacy.load('en_core_web_sm')
```

```python
sent=nlp('''Mark Zukerberg will meet Aditya Joshi in NEW york ,USA on monday 21,2024 4pm for $3 Trillion deal''')
```

```python
sent
```

```python
sent.ents # will print entities
```

```python
type(sent.ents)
```

```python
for word in sent.ents:
  print(word.ents[0],'-->',word.label_)
```

```python
spacy.explain('PERSON')
```
