```python
text= '''The tiger is a wild animal and the national animal of India. It is a very strong and beautiful animal with orange fur and black stripes. Tigers live in forests and hunt animals for food. They are good swimmers and can run very fast.

Tigers are carnivores, which means they eat meat. They mostly hunt deer and other small animals. A tiger usually lives alone and protects its area from other animals.'''
print(text)
```

```python
from sumy.summarizers.text_rank import TextRankSummarizer
```

```python
# import parser and tokenizer
from sumy.parsers.plaintext import PlaintextParser
from sumy.nlp.tokenizers import Tokenizer

```

```python
from nltk.tokenize import sent_tokenize
import nltk
nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('averaged_perceptron_tagger')
```

```python
snd=sent_tokenize(text)
print(snd)
```

```python
len(sent_tokenize(text))
```

```python
# initialize the parser
my_parser = PlaintextParser.from_string(text, Tokenizer("english"))
```

```python
# create the object of TextRank
text_rank_summarizer = TextRankSummarizer()
```

```python
summary= text_rank_summarizer(my_parser.document , sentences_count=3)
for sentence in summary:
    print(sentence,'\n')
```
